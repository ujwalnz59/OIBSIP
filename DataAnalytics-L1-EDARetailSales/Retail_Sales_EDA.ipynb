**Data Exploration (EDA)**

***Graph 1: Total Sales by category***

```python
import matplotlib.pyplot as plt

# Calculate total sales by category
category_sales = df.groupby("Category")["Sales"].sum()

# Create the bar chart
plt.figure(figsize=(8,5))
category_sales.plot(kind="bar", color=["skyblue","orange","green"])

# Add title and labels
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

# Display the chart
plt.show()
```

**Top 10 Best-Selling Products**

***Calculate Sales by products***

```python
# Calculate total sales for each product
top_products = df.groupby("Product Name")["Sales"].sum()

# Display the top 10 products
top_products = top_products.sort_values(ascending=False).head(10)

print(top_products)
```

***Create the Graph***

```python
plt.figure(figsize=(12,6))

top_products.plot(kind="bar", color="teal")

plt.title("Top 10 Best Selling Products")
plt.xlabel("Product Name")
plt.ylabel("Total Sales")
plt.xticks(rotation=75)

plt.show()
```

**Monthly sales Trend**

```python
# Convert Order Date from text to date format
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
# Create Year-Month column
df["Year-Month"] = df["Order Date"].dt.to_period("M")

# Calculate monthly sales
monthly_sales = df.groupby("Year-Month")["Sales"].sum()

# Convert period to string for plotting
monthly_sales.index = monthly_sales.index.astype(str)

# Create graph
plt.figure(figsize=(14, 6))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

**Quarterly Sales Trend**

***Create Quarter***

```python
# Create a Year-Quarter column
df["Year-Quarter"] = df["Order Date"].dt.to_period("Q")

# Calculate total sales for each quarter
quarterly_sales = df.groupby("Year-Quarter")["Sales"].sum()

print(quarterly_sales)
```

```python
# Create a Year-Quarter column
df["Year-Quarter"] = df["Order Date"].dt.to_period("Q")

# Calculate total sales for each quarter
quarterly_sales = df.groupby("Year-Quarter")["Sales"].sum()

# Convert quarter labels to text
quarterly_sales.index = quarterly_sales.index.astype(str)

plt.figure(figsize=(12, 6))

plt.plot(
    quarterly_sales.index,
    quarterly_sales.values,
    marker="o"
)

plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

**Correlation Heatmap**

```python
numeric_data = df[["Sales", "Quantity", "Discount", "Profit"]]

numeric_data.head()
```

```python
correlation = numeric_data.corr()

print(correlation)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

**Additional Visualization - Profit by Category**

```python
# Calculate total profit by category
category_profit = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)

# Create bar chart
plt.figure(figsize=(8, 5))

category_profit.plot(kind="bar")

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

**Before the Final Analysis - one useful analysis**

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit",
    alpha=0.6
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.tight_layout()
plt.show()
```

#*Conclusion*



The Exploratory Data Analysis of the Superstore dataset provided useful insights into overall sales and profit performance. Technology was the highest-performing category in terms of both sales and profit, while Furniture generated comparatively low profit despite having high sales. The monthly and quarterly analysis showed that sales generally increased over time, with the strongest performance occurring in the final quarter of 2017. The analysis also showed that higher discounts were associated with lower profits and more loss-making transactions. Overall, the findings suggest that the company should focus on high-performing products, carefully manage discounts, improve the profitability of the Furniture category, and plan inventory and promotions according to seasonal sales trends.
